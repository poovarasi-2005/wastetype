```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.tree import DecisionTreeClassifier
import joblib
```

```python
data=pd.read_csv("/content/waste_classification_dataset_500.csv")
print(data.head())
```

```python
x=data.drop("WasteType",axis=1)
y=data["WasteType"]
```

```python
encoder=LabelEncoder()
y=encoder.fit_transform(y)
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LogisticRegression()
lr.fit(X_train,y_train)
lr_pred=lr.predict(X_test)
lr_pred
```

```python
print("logistic regression accuracy")
print(accuracy_score(y_test,lr_pred))
print("/nClassification report")
print(classification_report(y_test,lr_pred))
print("/nConfusion matrix")
print(confusion_matrix(y_test,lr_pred))
```

```python
joblib.dump(lr,'Waste Type Classification_logistic regression.pkl')
```

```python
dt=DecisionTreeClassifier(max_depth=2)
dt.fit(X_train,y_train)
dt_pred=dt.predict(X_test)
```

```python
print("Decision tree accuracy")
print(accuracy_score(y_test,dt_pred))
print("/nClassification report")
print(classification_report(y_test,dt_pred))
print("/nConfusion matrix")
print(confusion_matrix(y_test,dt_pred))
```

```python
joblib.dump(dt,'Waste Type Classification_Decision tree.pkl')
```
